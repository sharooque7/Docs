```python
board = [["A","B","C","E"],["S","F","C","S"],["A","D","E","E"]]
visited_board = []
word = "SEE"
ROWS = len(board)
COLS = len(board[0])


for i in range(ROWS):
    rows = []
    for j in range(COLS):
        rows.append(False)
    visited_board.append(rows)
    
# print(visited_board)

def search_word():

    def dfs(i, j , k):
        if k == len(word):
            return True
        
        if i < 0 or j < 0 or i >= ROWS or j >= COLS or word[k] != board[i][j] or visited_board[i][j]:
            return False
        
        visited_board[i][j] = True

        res = dfs(i + 1, j , k +1) or \
        dfs(i - 1, j , k + 1) or \
        dfs(i, j+ 1, k + 1) or \
        dfs(i , j - 1, k + 1)

        visited_board[i][j] = False

        return res


    for i in range(ROWS):
        for j in range(COLS):
            if dfs(i, j, 0):
                return True


print(search_word())
```
